```python
def swap_numbers(a, b):
    return b, a

def is_even_or_odd(num):
    return "Even" if num % 2 == 0 else "Odd"

def largest_of_three(a, b, c):
    return max(a, b, c)

def is_palindrome(s):
    return s == s[::-1]

def fibonacci_series(n):
    fib_series = [0, 1]
    for i in range(2, n):
        fib_series.append(fib_series[-1] + fib_series[-2])
    return fib_series[:n]

def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

def reverse_string(s):
    return s[::-1]

def count_vowels(s):
    vowels = "aeiouAEIOU"
    return sum(1 for char in s if char in vowels)

def multiplication_table(n):
    return [n * i for i in range(1, 11)]

def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

def main():
    while True:
        print("\n--- Mini Toolkit ---")
        print("1. Swap two numbers")
        print("2. Check if a number is even or odd")
        print("3. Find the largest of three numbers")
        print("4. Check if a string is a palindrome")
        print("5. Print the Fibonacci series")
        print("6. Find factorial of a number (recursion)")
        print("7. Reverse a string (manual)")
        print("8. Count vowels in a string")
        print("9. Print multiplication table")
        print("10. Check if a number is prime")
        print("0. Exit")

        choice = input("Select an option (0-10): ")

        if choice == '1':
            a = float(input("Enter first number: "))
            b = float(input("Enter second number: "))
            a, b = swap_numbers(a, b)
            print(f"Swapped numbers: {a}, {b}")

        elif choice == '2':
            num = int(input("Enter a number: "))
            print(f"The number {num} is {is_even_or_odd(num)}.")

        elif choice == '3':
            a = float(input("Enter first number: "))
            b = float(input("Enter second number: "))
            c = float(input("Enter third number: "))
            print(f"The largest number is: {largest_of_three(a, b, c)}")

        elif choice == '4':
            s = input("Enter a string: ")
            print(f"The string '{s}' is a palindrome: {is_palindrome(s)}")

        elif choice == '5':
            n = int(input("Enter the number of terms in Fibonacci series: "))
            print("Fibonacci series:", fibonacci_series(n))

        elif choice == '6':
            n = int(input("Enter a number to find its factorial: "))
            print(f"The factorial of {n} is: {factorial(n)}")

        elif choice == '7':
            s = input("Enter a string to reverse: ")
            print("Reversed string:", reverse_string(s))

        elif choice == '8':
            s = input("Enter a string: ")
            print(f"The number of vowels in '{s}' is: {count_vowels(s)}")

        elif choice == '9':
            n = int(input("Enter a number for multiplication table: "))
            print("Multiplication table:")
            for i in multiplication_table(n):
                print(i)

        elif choice == '10':
            num = int(input("Enter a number to check if it's prime: "))
            print(f"The number {num} is prime: {is_prime(num)}")

        elif choice == '0':
            print("Exiting the toolkit. Goodbye!")
            break

        else:
            print("Invalid choice. Please select a valid option.")

if __name__ == "__main__":
    main()
```
